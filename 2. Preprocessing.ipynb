```python
# DATE-TIME FEATURE ENGINEERING
# Convert to datetime
df["hour_window"] = pd.to_datetime(df["hour_window"])
# Extract Date-Time Features
df["year"] = df["hour_window"].dt.year
df["month"] = df["hour_window"].dt.month
df["day"] = df["hour_window"].dt.day
df["hour"] = df["hour_window"].dt.hour
df["minute"] = df["hour_window"].dt.minute
# Monday = 0 .... Sunday = 6
df["weekday"] = df["hour_window"].dt.dayofweek
# Weekend Feature
df["is_weekend"] = df["weekday"].isin([5, 6]).astype(int)
# Working Hours (08:00 AM - 06:00 PM)
df["working_hours"] = (
    (df["hour"] >= 8) &
    (df["hour"] <= 18)
).astype(int)# CHECK NEW FEATURES
print("="*80)
print("NEW FEATURES CREATED")
print("="*80)
new_features = [
    "hour_window",
    "year",
    "month",
    "day",
    "hour",
    "minute",
    "weekday",
    "is_weekend",
    "working_hours"
]
display(df[new_features].head(10))
print("\nCurrent Shape :", df.shape)
```

```python
# REMOVE NON-INFORMATIVE FEATURES
df.drop(columns=["year", "minute"], inplace=True)
print("="*80)
print("UPDATED DATASET")
print("="*80)
print("Shape :", df.shape)
print("\nColumns:")
print(df.columns.tolist())
```

```python
# RULE-BASED SECURITY SCORE ENGINE
import numpy as np
# Initial Security Score
df["security_score"] = 100.0
# RULE 1 : Outside Working Hours
df.loc[df["working_hours"] == 0, "security_score"] -= 15
# RULE 2 : Weekend Activity
df.loc[df["is_weekend"] == 1, "security_score"] -= 10
# RULE 3 : Login Behaviour
df.loc[df["login_events"] >= 2, "security_score"] -= 5
df.loc[df["login_events"] >= 4, "security_score"] -= 10
df.loc[df["login_events"] >= 6, "security_score"] -= 20
# RULE 4 : Device (USB) Activity
df.loc[df["device_events"] >= 1, "security_score"] -= 10
df.loc[df["device_events"] >= 3, "security_score"] -= 15
df.loc[df["device_events"] >= 6, "security_score"] -= 25
# RULE 5 : File Activity
df.loc[df["file_events"] >= 3, "security_score"] -= 10
df.loc[df["file_events"] >= 8, "security_score"] -= 15
df.loc[df["file_events"] >= 15, "security_score"] -= 25
# RULE 6 : Email Activity
df.loc[df["email_events"] >= 5, "security_score"] -= 5
df.loc[df["email_events"] >= 10, "security_score"] -= 10
df.loc[df["email_events"] >= 20, "security_score"] -= 20
# RULE 7 : Attachments
df.loc[df["total_attachments"] >= 1, "security_score"] -= 5
df.loc[df["total_attachments"] >= 3, "security_score"] -= 10
df.loc[df["total_attachments"] >= 6, "security_score"] -= 20
# RULE 8 : Email Size
df.loc[df["total_email_size"] > 100000, "security_score"] -= 5
df.loc[df["total_email_size"] > 1000000, "security_score"] -= 10
df.loc[df["total_email_size"] > 10000000, "security_score"] -= 20
# RULE 9 : HTTPS Activity
df.loc[df["https_events"] >= 10, "security_score"] -= 5
df.loc[df["https_events"] >= 25, "security_score"] -= 10
df.loc[df["https_events"] >= 50, "security_score"] -= 20
# RULE 10 : Psychometric Adjustment
# High Conscientiousness increases security
df.loc[df["C"] >= 40, "security_score"] += 5
# Low Conscientiousness decreases security
df.loc[df["C"] <= 20, "security_score"] -= 5
# High Neuroticism decreases security
df.loc[df["N"] >= 40, "security_score"] -= 5
# Low Agreeableness decreases security
df.loc[df["A"] <= 20, "security_score"] -= 5
# COMPOUND RULES
# USB + File Activity
mask = (df["device_events"] >= 3) & (df["file_events"] >= 10)
df.loc[mask, "security_score"] -= 20
# Email + Attachments
mask = (df["email_events"] >= 15) & (df["total_attachments"] >= 3)
df.loc[mask, "security_score"] -= 15
# Weekend + Login
mask = (df["is_weekend"] == 1) & (df["login_events"] >= 2)
df.loc[mask, "security_score"] -= 15
# Outside Working Hours + HTTPS
mask = (df["working_hours"] == 0) & (df["https_events"] >= 20)
df.loc[mask, "security_score"] -= 15
# Outside Working Hours + File
mask = (df["working_hours"] == 0) & (df["file_events"] >= 5)
df.loc[mask, "security_score"] -= 20
# USB + Outside Working Hours
mask = (df["working_hours"] == 0) & (df["device_events"] >= 2)
df.loc[mask, "security_score"] -= 25
# Heavy Email + Large Email Size
mask = (df["email_events"] >= 20) & (df["total_email_size"] >= 1000000)
df.loc[mask, "security_score"] -= 20
# Clip Score
df["security_score"] = df["security_score"].clip(0,100)
# Security Status
conditions = [
    df["security_score"] >= 85,
    (df["security_score"] >= 65) &
    (df["security_score"] < 85),
    (df["security_score"] >= 40) &
    (df["security_score"] < 65),
    df["security_score"] < 40
]
choices = [
    "Secure",
    "Suspicious",
    "Unsecure",
    "Breached"
]
df["security_status"] = np.select(
    conditions,
    choices,
    default="Secure"
)
# Labels
label_map = {
    "Secure":0,
    "Suspicious":1,
    "Unsecure":2,
    "Breached":3

}
df["label"] = df["security_status"].map(label_map)
# Results
print("="*80)
print("SECURITY STATUS DISTRIBUTION")
print("="*80)
status = df["security_status"].value_counts().reset_index()
status.columns = ["Security Status","Records"]
status["Percentage"] = (status["Records"]/len(df)*100).round(2)
display(status)
print("\nSecurity Score Summary\n")
display(df["security_score"].describe())
display(df[[
    "security_score",
    "security_status",
    "label"
]].head(20))
```
